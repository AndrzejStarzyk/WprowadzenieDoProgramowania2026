ZADANIE: Wczytaj liczbę n i oblicz n-ty wyraz ciągu f(n) zadanego wzorem: f(0) == 1, f(n) == 3*f(n-1)^2 - 5*f(n-1) + 2. W tym celu wczytaj n - ilość iteracji pętli, które trzeba będzie wykonać, i zdefiniuj zmienną f przypisując 0.
```python
n = int(input())

f = 1

for _ in range(n):
    result = 3 * f * f - 5 * f + 2
    f = result
    print(f)
```

# Rekurencja

Oprócz dedykowanej instrukcji do wykonywania kodu w pętli jest jeszcze inny sposób , by osiągnąć ten sam efekt. Mianowicie zdefiniowanie funkcji i jako ostatnią instrukcję wywołanie tej samej funkcji. W ten sposób funkcja wykona się jeszcze raz od początku i tak w kółko aż nie napotka wyjścia funkcji albo przepełni się pamięć. 

Istotne informacje:

1. Przy każdym wywołaniu funkcji w pamięci odkładane są informacja o wywołaniu funkcji i jej argumentach. Ta pamięć zostanie zwolniona dopiero gdy funkcja skończy działanie, co w przypadku głębokiej rekurencji może długo potrwać.
2. Teoretycznie takżdy program napisany za pomocą rekurencji można rozwiązać przy użyciu pętli, i także na odwrót. Jednak praktycznie wykorzystanie iteracji jest lepsze - kod jest czytelniejszy, łatwiej przewidzieć co się wydarzy, czy i kiedy dana linijka kodu się wykona. Z drugiej strony wiele istotnych algorytmów i metod programowania polega na rekurencji ("dziel i rządź", programowanie dynamiczne, sortowanie szybkie, itp.)


```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

# Postawowe struktury danych

## Lista

Struktura danych przechowująca obiekty tego samego typu, na przykład integer-y, float-y, inne listy, itp. Dane są ustawione w zadanej kolejności i ponumerowane do 0 w górę. 

W informatyce wyróżniamy tablicę i listę. Ta pierwsza polega na zarezerwowaniu w pamięci miejsca na sekwencję elementów o tym samym typie (tym samym rozmiarze). W ten sposób łatwo odnaleźć każdy element nawet bardzo długiej list, ale trudno dodawać i usuwać elementy, bo trzeba robić niekiedy skomplikowane przenoszenia w pamięci. List w informatyce polega na szeregu obiektów, z których każdy przechowuję jakąś wartość oraz wskazanie do poprzedniego i następnego obiektu (symbolizujących odpowiednio poprzedni i następny element listy). Takie obiekty mogą być w różnych miejscach w pamięci, ale dzięki ty wskaźnikom łączą się w listę. W ten łatwo dodawać i usuwać elementy, bo wystarczy odpowiednio pozamieniać te wskaźniki (bez przenoszenia niczego w pamięci). Za to trudno znaleźć dany element, bo trzeba po kolei sprawdzać każdy element.

Lista w Pythonie łączy cechy obu tych konstrukcji i bez poroblemu można odczytywać oraz manipulować elementami listy.

```python
numbers = [1, 2, 3, 4, 5]
```

### Operacje na liście

```python
numbers = [1, 2, 3, 4, 5]

numbers[0] # odczytanie pierwszego elementu listy

numbers.append(6) # dodanie elementu na końcu listy

last = numbers.pop() # usunięcie ostatniego elementu listy i zwrócenie go

numbers.insert(2, 10) # wstawienie elementu na pozycji 2

numbers.remove(3) # usunięcie pierwszego wystąpienia elementu 3

numbers[3] = 20 # zmiana wartości elementu na pozycji 3

len(numbers) # zwraca długość listy

numbers.reverse() # odwraca kolejność elementów w liście

numbers.sort(key=None, reverse=False) # sortuje elementy w liście

sorted_list = sorted(numbers, key=None, reverse=False) # zwraca nową posortowaną listę, nie zmieniając oryginalnej

new_list = numbers.copy() # zwraca kopię listy

numbers + new_list # konkatenacja dwóch list

numbers * 2 # powielenie listy

1 in numbers # sprawdzenie, czy element 1 znajduje się w liście
```

```python
list_of_lists = [[1, 2], [3, 4], [5, 6]] # lista list

list_of_lists[1][0] # odczytanie pierwszego elementu drugiej listy w liście list

len(list_of_lists) # zwraca długość listy list
len(list_of_lists[0]) # zwraca długość pierwszej listy w liście list
```

```python
list_of_numbers = [1, 2, 3, 4, 5]

sum_of_numbers = sum(list_of_numbers) # zwraca sumę elementów w liście

maximum = max(list_of_numbers) # zwraca największy element w liście

minimum = min(list_of_numbers) # zwraca najmniejszy element w liście
```

### Slicing

Metoda odczytywania dowolnego elementu lub sekwencji elementów z list. Wynik działania takiej operacji to nowa lista albo jej element

```python
numbers = [1, 2, 3, 4, 5]

numbers[0] # odczytanie pierwszego elementu listy

numbers[3] # odczytanie czwartego elementu listy (indeks 3)

numbers[-1] # odczytanie ostatniego elementu listy

numbers[-3] # odczytanie trzeciego elementu od końca listy (indeks -3)

numbers[1:4] # odczytanie elementów od indeksu 1 do 3 (indeks 4 nie jest wliczany)

numbers[:3] # odczytanie pierwszych trzech elementów listy (indeks 3 nie jest wliczany)

numbers[2:] # odczytanie elementów od indeksu 2 do końca listy

numbers[2:-1] # odczytanie elementów od indeksu 2 do przedostatniego elementu listy

numbers[:] # odczytanie wszystkich elementów listy (tworzy kopię listy)

numbers[::2] # odczytanie co drugiego elementu listy (indeksy 0, 2, 4, ...)
```

## Krotka (tuple)

Struktura danych przechowująca w zadanej kolejność, określoną liczbę obiektów, niekonieczne o tym samym typie. Moża wykonać operacje slicingu, ale nie da się wykonać żadnej manipulacji elementami - raz stworzona krotka nie może zostać zmieniona (niemutowalna, immutable) - trzeba ewentualnie stworzyć nową.

```python
ones = (1, 1.0, "1", True, [1, 2, 3], {1, 2, 3}) # krotka z różnymi typami danych

ones[0] # odczytanie pierwszego elementu krotki

len(ones) # zwraca długość krotki
```

# Zbiór (set)

Stuktura danych przechowująca obiekty niekonieczne o tym samym typie, bez zachowania ich kolejność. 

```python
set_of_numbers = {1, 2, 3, 4, 5} # zbiór liczb

set_of_numbers.add(6) # dodanie elementu do zbioru

set_of_numbers.remove(3) # usunięcie elementu ze zbioru

set_of_numbers.discard(10) # usunięcie elementu ze zbioru, jeśli istnieje (nie powoduje błędu, jeśli element nie istnieje)

len(set_of_numbers) # zwraca liczbę elementów w zbiorze

1 in set_of_numbers # sprawdzenie, czy element należy do zbioru

111 not in set_of_numbers # sprawdzenie, czy element nie należy do zbioru

set_of_numbers.clear() # usuwa wszystkie elementy ze zbioru

set_of_numbers = set_of_numers2 = {1, 2, 3, 4, 5}

sum_of_sets = set_of_numbers.union(set_of_numers2) # suma zbiorów

intersection_of_sets = set_of_numbers.intersection(set_of_numers2) # część wspólna zbiorów
```

# Słownik (dict)

Struktura danych przechowująca dane w postaci (klucz, warotść). Zamiast w liście na konkretnych pozycjach umieszczać dane (przypisywać liczbę w kolejność), w słowniku mamy klucze jednoznacznie identyfikujące dane. Nie ma narzuconej kolejność, wartości i klucze mogą mieć różne typy, ale klucze muszą być immutable i unikalne w słowniku

```python
dictionary = {"a": 1, "b": 2, "c": 3} # słownik z kluczami i wartościami

dictionary["a"] # odczytanie wartości dla klucza "a"

dictionary["d"] = 4 # dodanie nowego elementu do słownika

dictionary["b"] = 5 # zmiana wartości dla klucza "b"

dictionary.pop("c") # usunięcie elementu ze słownika dla klucza "c"

dictionary.get("b") # odczytanie wartości dla klucza "b" (zwraca None, jeśli klucz nie istnieje)

dictionary.get("e", 5) # odczytanie wartości dla klucza "e" (zwraca 5, jeśli klucz nie istnieje)

dictionary.keys() # zwraca listę kluczy w słowniku

dictionary.values() # zwraca listę wartości w słowniku

dictionary.items() # zwraca listę par (klucz, wartość) w słowniku

```

## String (str)

```python
text = "Hello, World!"

text[0] # odczytanie pierwszego znaku w stringu

text[2:5] # odczytanie znaków od indeksu 2 do 4 (indeks 5 nie jest wliczany)

text.capitalize() # zwraca string z pierwszą wielką literą

text.upper() # zwraca string z wszystkimi wielkimi literami

text.lower() # zwraca string z wszystkimi małymi literami

text.split(",") # dzieli string na listę elementów, używając przecinka jako separatora

text.replace("World", "Python") # zwraca string z zamienionym słowem "World" na "Python"

text.strip() # zwraca string bez białych znaków na początku i końcu

text.startswith("Hello") # sprawdza, czy string zaczyna się od "Hello"
```

## Comprehension

```python
[x for x in range(10)] # lista liczb od 0 do 9

[0 for _ in range(10)] # lista dziesięciu zer

[x for x in range(10) if x % 2 == 0] # lista liczb parzystych od 0 do 9

[[(i, j) for j in range(i, i+3)] for i in range(5)] # lista list współrzędnych (i, j) dla i w zakresie 0-4 i j w zakresie i do i+2

{x: x**2 for x in range(5)} # słownik z kluczami jako liczby od 0 do 4 i wartościami jako kwadraty tych liczb
```
