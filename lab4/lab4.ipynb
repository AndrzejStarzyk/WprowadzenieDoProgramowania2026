# Zadania

Zadanie 1: Napisz funkcję z jednym parametrem - listą liczb całkowitych. Funkcja powinna zwrócić największy element na tej liście. (nie używaj funkcji max())

Zadanie 2: Napisz funkcję z jednym parametrem - listą liczb całkowitych. Funkcja powinna zwrócić najmniejszy element na tej liście. (nie używaj funkcji min())

Zadanie 3: Napisz funkcję z dwoma parametrami - listą liczb całkowitych i liczbą całkowitą 'k'. Funkcja powinna zwrócić True jeżeli k znajduje się na tej liście i False w przeciwnym wypadku. (nie używaj operatora 'in')

Zadanie 4: Napisz funkcję z jednym parametrem - listą 'to_reverse'. Funkcja powinna zwrócić listę, której elementy są odrotnej kolejnoście w porównaniu z listą 'to_reverse'. (nie używaj funkcji wobudowanych ani z bibliotek, które odwracają listę).

Zadanie 5: Napisz funkcję z dwoma parametrami - listą liczb całkowitych i liczbą całkowitą 'k'. Funkcja powinna zwrócić True jeżeli suma pewych dwóch elementów listy równa się k i False w przeciwnym wypadku.

 Zadanie 6: Napisz funkcję z jednym parametrem - liczbą całkowitą nieujemną 'k'. Funkcja powinna zwrócić True jeżeli ta liczba jest pierwsza i False w przeciwnym wypdaku. 
 Wskazówka: Należy po kolei, zaczynając od 1, sprawdzać, podzielność zadanej liczby przez mniejsze liczby. Nie ma sensu sprawdzać podzielności powyżej pierwiastka z 'k', gdyż dzielniki zawsze są w parach jeden większy drugi mniejszy od pierwiastka (lub równy).
 Istnieje jeszcze kilka innych usprawnień pozwalających nie sprawdzać wszystkich liczb po kolei, zachęcam do zastanowienia się na tym.

Zadanie 7: Napisz funkcję z jednym parametrem - listą liczb 'to_sort'. Funkcja powinna zwrócić nową listę z tymi samymi liczbami w rosnącej kolejności. W tym celu stwórz nową listę 'result' i w pętli usuwaj najmniejszy pozostały element w 'to_sort' i dodawaj go do 'result'.

Zadanie 8: Napisz funkcję z jednym parametrem - tablicą binarną (lista list, gdzie elementy list to albo 0, albo 1). Funkcja powinna dla każdej komórki policzyć w ilu sąsiednich komórkach jest 1. Funkcja powinna zwrócić nową tablicę - każda jej komórka powinna zawierać ilość sąsiednich komórek na tablicy binarnej, które zawierają 1. (Potrzebne będą dwie pętle, trzeba sprawdzać, czy sąsiednie komórki nie są przypadkiem poza tablicą)
